```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

data1 = sns.load_dataset('tips')
df1 = pd.DataFrame(data1)
data2 = px.data.gapminder()
df2 = pd.DataFrame(data2)
```

```python
#                               line chart
temp_data = df2[df2["country"].isin(["India", "Brazil", "Germany"])]
sns.lineplot(data = temp_data, x = "year", y = "lifeExp", hue = "country", style = "continent", markers = True, palette = "magma")
plt.show()
```

```python
#                            count plot chart
sns.countplot(data = df1, x = "sex", hue = "smoker", palette = "magma", order = df1["sex"].value_counts(ascending = True).index, alpha = 0.8)
plt.show()
```

```python
#                               bar chart
sns.barplot(data = df1, x = "sex", y = "total_bill", hue = "smoker", palette = ["green", "blue"], orient = "v", estimator = np.mean)
plt.show()
```

```python
#                        histogram chart
sns.histplot(data = df1, x = "total_bill", bins = 5, hue ="sex", palette = ["black", "red"], kde = True)

plt.show()
```

```python
#                         scatterplot chart
sns.scatterplot(data = df1, x = "total_bill", y = "tip", hue = "sex", style = "smoker", size = "total_bill", palette = ["green", "purple"])
plt.show()
```

```python
#                          kdeplot chart
sns.kdeplot(data = df1, x = "total_bill", hue = "sex", palette = [ "green", "blue"], fill = True, alpha = 0.4)
plt.show()
```

```python
#                             boxplot chart
sns.boxplot(data = df1, y = "total_bill", x = "smoker", hue = "sex", palette = ["yellow", "purple"], orient = "v")
plt.show()
```

```python
#                       heatmap chart
data = df2[df2["continent"] == "Europe"]
temp_data = data.pivot(index = "country", columns = "year", values = "lifeExp")
plt.figure(figsize = (15, 15))
sns.heatmap(data = temp_data, annot = True, cmap = "coolwarm", fmt = ".2f", linewidth = 0.5, linecolor = "black")
plt.show()
```

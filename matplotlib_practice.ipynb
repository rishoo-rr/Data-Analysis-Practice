```python
import matplotlib.pyplot as plt
#              line  chart
x1 = [1,2,3,4,5]
x2 = [3,4,5,6,7]
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.figure(figsize = (4,3))
plt.plot(x1, y1, color = "green", marker = "o", linestyle = "--", linewidth = 2, markersize = 8, label = ("sales 2024"))
plt.plot(x2, y2, color = "blue", marker = "o", linestyle = "--", linewidth = 2, markersize = 8, label = ("sales 2025"))
plt.title("year on year sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
#                bar chart
x = ["A", "B", "C", "D", "E"]
y = [30, 25, 60, 20, 90]

plt.figure(figsize = (4,3))
plt.bar(x,y, width = 0.8, color = ["green", "blue"], edgecolor = "black", linewidth = 2, tick_label = ["jan", "feb", "mar", "apr", "may" ])
plt.title("sales of 5 months")
plt.xlabel("months")
plt.ylabel("month on month sales")
plt.yticks(ticks = [30, 60, 90], labels = ["bad", "good", "best"])
plt.show() 
```

```python
#                 histogram chart
data = [20, 19, 27, 67, 35, 40, 45, 55, 43, 83, 93, 57, 64, 78, 88, 21, 75, 30, 65, 95, 15, 50, 70, 34]
plt.figure(figsize = (4,3))
plt.hist(data, bins = 5, color = "gray", edgecolor = "black")
plt.title("marks distribution")
plt.xlabel("marks scored")
plt.ylabel("number of students")
plt.show()
```

```python
#                  pie chart
x = ["A", "B", "C", "D", "E"]
y = [30, 25, 60, 20, 90]

plt.figure(figsize = (4,3))
plt.pie(y, labels = x, autopct = "% 1.1f%%", startangle = 90, explode = [0, 0, 0, 0, 0.1],  colors = ["purple", "gray", "blue"], shadow =True)
plt.show()
```

```python
#                     scatter chart
study_hours = [2, 4, 5, 7, 8]
marks = [40, 55, 65, 80, 95]

plt.figure(figsize = (4,3))
plt.scatter(study_hours, marks, color = "skyblue",s = 50, marker = "*")
plt.title("relation anylises")
plt.xlabel("study_hours")
plt.ylabel("marks")
plt.show()
```

```python
#              subplot 
plt.figure(figsize = (12,4))
#              line  chart
plt.subplot(1, 3, 1)
x1 = [1,2,3,4,5]
x2 = [3,4,5,6,7]
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.title("year on year sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.plot(x1, y1, color = "gray", marker = "o", linestyle = "--", linewidth = 2, markersize = 8, label = ("sales 2024"))
plt.plot(x2, y2, color = "skyblue", marker = "o", linestyle = "--", linewidth = 2, markersize = 8, label = ("sales 2025"))
plt.legend()

#                bar chart
plt.subplot(1, 3 ,2)
x = ["A", "B", "C", "D", "E"]
y = [30, 25, 60, 20, 90]

plt.bar(x,y, width = 0.8, color = ["purple", "skyblue"], edgecolor = "black", linewidth = 2, tick_label = ["jan", "feb", "mar", "apr", "may" ])
plt.title("sales of 5 months")
plt.xlabel("months")
plt.ylabel("month on month sales")
plt.yticks(ticks = [30, 60, 90], labels = ["bad", "good", "best"])

#                   pie chart
plt.subplot(1, 3, 3)
x = ["A", "B", "C", "D", "E"]
y = [30, 25, 60, 20, 90]

plt.pie(y, labels = x, autopct = "% 1.1f%%", startangle = 90, explode = [0, 0, 0, 0, 0.1],  colors = ["yellow", "gray", "blue"], shadow =True)
plt.show()
```
